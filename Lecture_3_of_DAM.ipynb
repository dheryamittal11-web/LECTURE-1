```python
import pandas as pd
import numpy as np

# Ensure df is defined by loading the dataset
df = pd.read_excel("/content/Lecture2_data.xlsx")
print(df.dtypes)
```

```python
#Top 10 profit/loss product
df["Total_Cost_Lakh"] = (
  +df["Marketing_Spend_Lakh"]
   + df["Content_Cost_Lakh"]
    +df["Salary_Cost_Lakh"]
    +df["Other_Cost_Lakh"]
)

df["Profit_Lakh"] = df["Revenue_Lakh"] - df["Total_Cost_Lakh"]

print(df[["Product", "Revenue_Lakh", "Total_Cost_Lakh", "Profit_Lakh"]].head(10))
```

```python
# Missing Value
print(df. isna(). sum())
```

```python
# Decision making structures
#If Statements
rating = df["Customer_Rating"].mean()

if rating >= 4:
  print("Customer satisfaction is good")
```

```python
revenue = df["Revenue_Lakh"].mean()

if revenue >= 300:
  print("Average revenue is high")
else:
  print("Average revenue is low")
```

```python
# Decision making structures
# Nested IF Statement in Python
revenue = df["Revenue_Lakh"].mean()
rating = df["Customer_Rating"].mean()

if revenue >= 300:
  if rating >= 4:
    print("Average revenue is high and customer satisfaction is good")
  else:
    print("Average revenue is high and customer satisfaction is bad")
else:
  print("Revenue is low")
```

```python
# Find the average customer rating
rating = df["Customer_Rating"].mean()

# Make a decision based on the rating
if rating >= 4.5:
  print("Execellent")

elif rating >= 4.0:
 print("Good")

elif rating >= 3.0:
  print("Average")

else:
  print("Needs Improvement")
```
